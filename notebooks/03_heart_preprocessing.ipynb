```python
import pandas as pd
import numpy as np
```

```python
try:
    df = pd.read_csv('../data/processed/heart_cleaned.csv')
    print("Successfully loaded 'heart_cleaned.csv'.")
    print("Initial shape:", df.shape)
except FileNotFoundError:
    print("Error: 'heart_cleaned.csv' not found. Please ensure the file from the first step is in the same directory.")
    exit()
```

```python

# --- Step 2.1: Target Variable Transformation ---
print("\n--- Step 2.1: Transforming Target Variable ---")

# Rename the 'num' column to 'target'
df.rename(columns={'num': 'target'}, inplace=True)
print("Renamed 'num' column to 'target'.")

# Convert the 'target' column to a binary format
# Values > 0 are mapped to 1 (heart disease present), 0 remains 0 (no disease)
df['target'] = (df['target'] > 0).astype(int)
print("Transformed 'target' to binary format (0 or 1).")
print("Target variable distribution:\n", df['target'].value_counts())
```

```python
# --- Step 2.2: Correction of Invalid Clinical Data ---
print("\n--- Step 2.2: Correcting Invalid Clinical Data ---")

# Replace 0 values in 'trestbps' and 'chol' with NaN as they are physiologically impossible
print(f"Values of 0 in 'trestbps' before correction: {(df['trestbps'] == 0).sum()}")
print(f"Values of 0 in 'chol' before correction: {(df['chol'] == 0).sum()}")

df['trestbps'].replace(0, np.nan, inplace=True)
df['chol'].replace(0, np.nan, inplace=True)

print("\nReplaced 0s with NaN in 'trestbps' and 'chol'.")
print(f"NaNs in 'trestbps' after correction: {df['trestbps'].isnull().sum()}")
print(f"NaNs in 'chol' after correction: {df['chol'].isnull().sum()}")
```

```python
# --- Step 2.3: Imputation of Missing Values ---
print("\n--- Step 2.3: Imputing Missing Values ---")

# Define categorical and numerical columns for imputation
categorical_cols_impute = ['fbs', 'restecg', 'exang', 'slope', 'thal']
numerical_cols_impute = ['trestbps', 'chol', 'thalch', 'oldpeak', 'ca']

# Impute categorical features with the mode
print("\nImputing categorical features with mode...")
for col in categorical_cols_impute:
    if col in df.columns:
        mode_val = df[col].mode()[0]
        df[col].fillna(mode_val, inplace=True)
        print(f"  - Filled NaNs in '{col}' with mode ('{mode_val}').")

# Impute numerical features with the median
print("\nImputing numerical features with median...")
for col in numerical_cols_impute:
    if col in df.columns:
        median_val = df[col].median()
        df[col].fillna(median_val, inplace=True)
        print(f"  - Filled NaNs in '{col}' with median ({median_val}).")

# Verify that there are no more missing values
print("\nMissing values count after imputation:\n", df.isnull().sum())
```

```python
# --- Step 2.4: Feature Engineering and Finalization ---
print("\n--- Step 2.4: Feature Engineering and Finalization ---")

# Drop unused columns
print("\nDropping 'id' and 'dataset' columns...")
df.drop(columns=['id', 'dataset'], inplace=True)
print("Columns dropped.")

# Define columns for one-hot encoding
categorical_cols_encode = ['sex', 'cp', 'fbs', 'restecg', 'exang', 'slope', 'thal']

# Perform one-hot encoding
print("\nPerforming one-hot encoding on categorical features...")
df = pd.get_dummies(df, columns=categorical_cols_encode, drop_first=True)
print("One-hot encoding complete.")
print("New shape of DataFrame:", df.shape)
print("New columns:", df.columns.tolist())

# Save the fully processed data to a new CSV file
output_filename = '../data/processed/heart_processed.csv'
df.to_csv(output_filename, index=False)
print(f"\nPreprocessing complete. The final dataset has been saved as '{output_filename}'.")
```
