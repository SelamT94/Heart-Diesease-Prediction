```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('../data/heart.csv')
```

```python
df['num'].unique()
```

the data orginally have 5 classes. 0 being not diseased and 1 up to 4 being different stages of heart disease. we only want to predict for Heart disease or no Heart dieases so we transform our target

```python
df.rename(columns={'num': 'target'}, inplace=True)

df['target'] = (df['target'] > 0).astype(int)
print("Target variable distribution:\n", df['target'].value_counts())
```

#### Some of the 0 values cant be 0 because it is physiologically impossible so we change them to NaN.

```python
df['trestbps'].replace(0, np.nan, inplace=True)
df['chol'].replace(0, np.nan, inplace=True)
```

### Imputing Missing Values

```python
categorical_cols_impute = df.select_dtypes(include=['object']).columns

print("Categorical features are imputed by the mode of the column.")
for col in categorical_cols_impute:
    if col in df.columns:
        mode_val = df[col].mode()[0]
        df[col].fillna(mode_val, inplace=True)
        print(f"Imputed NaNs in '{col}' with mode ('{mode_val}').")
```

```python
numerical_cols_impute = df.select_dtypes(include=np.number).columns

print("Numerical features are imputed by the median of the column")
for col in numerical_cols_impute:
    if col in df.columns:
        median_val = df[col].median()
        df[col].fillna(median_val, inplace=True)
        print(f"Imputed NaNs in '{col}' with median ({median_val}).")
```

```python
print("\nMissing values count after imputation:", df.isnull().sum())
```

```python
df.drop(columns=['id', 'dataset'], inplace=True)
```

### Feature Engineering

```python

categorical_cols_encode = df.select_dtypes(include=['object']).columns

df = pd.get_dummies(df, columns=categorical_cols_encode, drop_first=True)
print("New shape of DataFrame:", df.shape)
print("New columns:", df.columns.tolist())

output_filename = '../data/heart_processed.csv'
df.to_csv(output_filename, index=False)
```
