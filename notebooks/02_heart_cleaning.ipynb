```python
import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_csv('../data/raw/heart.csv')
```

```python
# --- 1. Initial Data Inspection ---
print("--- Initial Data Inspection ---")

# Display the first 5 rows
print("\n1. First 5 Rows of the Dataset:")
print(df.head())
```

```python
# Display concise summary of the DataFrame
print("\n2. Dataframe Info:")
df.info()
```

```python
# Display descriptive statistics for numerical columns
print("\n3. Descriptive Statistics:")
print(df.describe())
```

```python
# --- 2. Data Cleaning ---
print("\n--- Data Cleaning ---")

# Check for missing values
print("\n1. Missing Values per Column:")
print(df.isnull().sum())

# Check for duplicate rows
duplicate_count = df.duplicated().sum()
print(f"\n2. Number of Duplicate Rows: {duplicate_count}")

# Remove duplicate rows
if duplicate_count > 0:
    df.drop_duplicates(inplace=True)
    print(f"\n   Removed {duplicate_count} duplicate rows.")
    print(f"   New shape of the DataFrame: {df.shape}")
```

```python
# --- 3. Deeper Data Understanding ---
print("\n--- Deeper Data Understanding ---")

# Identify and inspect categorical and numerical features
categorical_cols = df.select_dtypes(include=['object', 'category']).columns
numerical_cols = df.select_dtypes(include=np.number).columns

print("\n1. Identified Column Types:")
print(f"   - Numerical Columns: {list(numerical_cols)}")
print(f"   - Categorical Columns: {list(categorical_cols)}")

# Analyze unique values in categorical columns to understand them better
# (Assuming 'sex', 'cp', 'fbs', 'restecg', 'exang', 'slope', 'ca', 'thal', 'target' are categorical in nature, even if numeric)
potential_categorical = ['sex', 'cp', 'fbs', 'restecg', 'exang', 'slope', 'ca', 'thal', 'target']
print("\n2. Unique Values in Potential Categorical Features:")
for col in potential_categorical:
    if col in df.columns:
        print(f"   - {col}: {df[col].unique()}")

# Analyze the distribution of the target variable
print("\n3. Target Variable Distribution:")
if 'target' in df.columns:
    print(df['target'].value_counts(normalize=True))
else:
    print("   'target' column not found.")

# Save the cleaned dataframe to a new CSV file for the next steps.
df.to_csv('../data/processed/heart_cleaned.csv', index=False)
```
